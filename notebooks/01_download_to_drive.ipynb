# 下载 Qwen3.8-27B-Uncensored (Q8_0 GGUF, 29GB) 到 Google Drive

下载不需要 GPU，用 CPU 运行时即可（不消耗计算单元）。如果 VS Code 里 `drive.mount` 失败，把这个 notebook 上传到网页版 Colab 运行。

```python
!pip install -q -U "huggingface_hub[hf_transfer]"
import os
os.environ["HF_HUB_ENABLE_HF_TRANSFER"] = "1"

from google.colab import drive
drive.mount('/content/drive')

REPO = "JonathanColetti/Qwen3.8-27B-Uncensored-GGUF"
FILE = "Qwen3.8-27B-Uncensored-Q8_0.gguf"   # 29.05 GB；想更小可换成 ...-Q6_K.gguf (22GB) / ...-Q4_K_M.gguf (17GB)
LOCAL = "/content/tmp"
DRIVE_DIR = "/content/drive/MyDrive/models/" + REPO.replace("/", "__")

# 检查空间：本地盘 & Drive
!df -h /content | tail -1
!df -h /content/drive | tail -1
```

```python
from huggingface_hub import hf_hub_download

# 先下到本地盘（快、稳），再复制到 Drive；中断后重跑会续传
p = hf_hub_download(REPO, FILE, local_dir=LOCAL)
hf_hub_download(REPO, "README.md", local_dir=LOCAL)
print("本地完成:", p)

!mkdir -p "{DRIVE_DIR}"
!rsync -ah --progress "{p}" "{LOCAL}/README.md" "{DRIVE_DIR}/"
!ls -lh "{DRIVE_DIR}"
```

```python
# 确保写入同步到 Drive 再断开（大文件可能要等几分钟）
drive.flush_and_unmount()
print("已同步到 Drive: MyDrive/models/JonathanColetti__Qwen3.8-27B-Uncensored-GGUF/")
```
